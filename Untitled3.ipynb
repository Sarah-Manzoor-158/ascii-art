```python
def is_fashionably_late(arrivals, name):
    """
    A guest is fashionably late if they arrived after at least half
    of the party's guests, but are not the very last guest.
    """
    order = arrivals.index(name)
    halfway_mark = (len(arrivals) + 1) // 2

    return order >= halfway_mark and order < len(arrivals) - 1

party_attendees = ['Adela', 'Fleda', 'Owen', 'May', 'Mona', 'Gilbert', 'Ford']
print(is_fashionably_late(party_attendees, 'Mona'))
print(is_fashionably_late(party_attendees, 'Gilbert'))
print(is_fashionably_late(party_attendees, 'Ford'))
print(is_fashionably_late(party_attendees, 'Adela'))
```
