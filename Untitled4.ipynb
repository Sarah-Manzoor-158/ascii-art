```python
def get_captain(teams):
    last_team = teams[-1]
    captain = last_team[1]
    return captain
teams_list = [
    ["Alice", "Bob", "Charlie"],
    ["Sarah", "Esha", "Farah"],
    ["Eve", "Ali", "Zainab"]
]

print(get_captain(teams_list))
```

```python
def same_meal(meals):
    for i in range(len(meals) - 1):
        if meals[i] == meals[i + 1]:
            return True
    return False
meal1 = ['Biryani', 'Pizza', 'Pizza', 'Burger']
meal2 = ['Pizza', 'Burger', 'Salad', 'Pasta']
print(same_meal(meal1))
print(same_meal(meal2))
```
